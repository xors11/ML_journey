```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[4,5,6,7]
#line plot
plt.plot(x,y)
plt.xlabel("X AXIS")
plt.ylabel('Y AXIS')
plt.title("BASIC PLOTTING")
plt.show()
```

```python
x=[1,2,3,5]
y=[1,3,5,7]
z=[4,6,9,1]

plt.figure(figsize=(9,5))
plt.subplot(2,2,1)
plt.plot(x,y,color='red',marker='o')
plt.title('X-Y')

plt.subplot(2,2,2)
plt.plot(y,z,color='green',marker='o')
plt.title("Y-Z")

plt.subplot(2,2,3)
plt.plot(z,x,color='yellow',marker='o')
plt.title('Z-X')


```
