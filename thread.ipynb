```python
import time
import threading
def print_numbers():
    for i in range(7):
        time.sleep(2)
        print(f"Numbers:",{i})

def print_alphabets():
    for x in "abcdefg":
        print(f"Alphabets:",{x})

t=time.time()
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_alphabets)
t1.start()
t2.start()
print(time.time()-t)
```
