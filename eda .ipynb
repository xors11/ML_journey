```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel('flight_price.xlsx')
df.head(2)
```

```python
df.info()
```

```python
df['Date']=df['Date_of_Journey'].str.split("/").str[0]
df['Month']=df['Date_of_Journey'].str.split("/").str[1]
df['Year']=df['Date_of_Journey'].str.split("/").str[2]
df.head(2)
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df["Year"]=df['Year'].astype(int)
df.info()
```

```python
df['Dep_hour']=df['Dep_Time'].str.split(':').str[0]
df['Dep_min']=df['Dep_Time'].str.split(":").str[1]
df['Dep_hour']=df['Dep_hour'].astype(int)
df['Dep_min']=df['Dep_min'].astype(int)
df.drop('Dep_Time',axis=1,inplace=True)
df.head(2)
```

```python
df.head(2)
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df["Total_Stops"].map({"non-stop": 0,
    "1 stop": 1,
    "2 stops": 2,
    "3 stops": 3,
    "4 stops": 4})
```

```python
df.head(2)
```

```python
df['Additional_Info'].unique()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
df1=pd.DataFrame(encoder.fit_transform(df[["Additional_Info"]]).toarray(),columns=encoder.get_feature_names_out())

```

```python
df.head(2)
```

```python
df2=pd.concat([df,df1],axis=1)
df2.head(2)
```

```python
df3=pd.DataFrame(encoder.fit_transform(df[["Airline","Source","Destination"]]).toarray(),columns=encoder.get_feature_names_out())
df4=pd.concat([df2,df3],axis=1)
df4.head(2)
```

```python
df4.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df4.head(2)
```

```python
df4['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df4['Arrival_hour']=df4['Arrival_Time'].str.split(":").str[0]
df4['Arrival_min']=df4['Arrival_Time'].str.split(":").str[1]
```

```python

df4.head(2)
```

```python
df4.drop("Arrival_Time",axis=1,inplace=True)
```

```python
df4.head(2)
```

```python
df4.drop("Additional_Info",axis=1,inplace=True)
```

```python
df4.head(2)
```

```python
df4["Duration_Hour"] = df4["Duration"].str.extract(r"(\d+)h").fillna(0).astype(int)

df4["Duration_Min"] = df4["Duration"].str.extract(r"(\d+)m").fillna(0).astype(int)
```

```python
df4.head(5)
```

```python
df4.drop('Duration',axis=1,inplace=True)
```

```python
df4.info()
```

```python
df4['Total_Duration']=df4['Duration_Hour'] * 60 + df4['Duration_Min'].astype(int)
```

```python
df4.head(2)
```

```python
df4.drop("Airline",axis=1,inplace=True)
```

```python
df4.drop("Duration_Hour",axis=1,inplace=True)
df4.drop("Duration_Min",axis=1,inplace=True)
```

```python
df4.drop("Source",axis=1,inplace=True)
df4.drop("Destination",axis=1,inplace=True)
```
