```python
import threading
import requests
from bs4 import BeautifulSoup
url=['https://www.udemy.com/course/complete-machine-learning-nlp-bootcamp-mlops-deployment/learn/lecture/44469918#overview',
    'https://supabase.com/dashboard/project/fetqsatawztrnqnoneqy/storage/analytics']
def fetch_content(url):
    response=requests.get(url)
    soup=BeautifulSoup(response.content,'html.parser')
    print(f"Fetched {len(response.content)} characters from {url}")
thread=[]
for urls in url:
    t1=threading.Thread(target=fetch_content,args=(urls,))
    thread.append(t1)
    t1.start()
    
```
